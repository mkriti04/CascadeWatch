```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.read_graphml("cascadewatch_graph.graphml")

pos = {node: (float(data["x"]), float(data["y"])) for node, data in G.nodes(data=True)}

colors = []
sizes = []
for node, data in G.nodes(data=True):
    node_type = data.get("node_type", "unknown")
    if node_type == "road":
        colors.append("gray")
        sizes.append(10)    # Small nodes for roads
    elif node_type == "hospital":
        colors.append("red")
        sizes.append(40)    # Larger nodes for hospitals
    elif node_type == "power":
        colors.append("blue")
        sizes.append(60)    # Largest nodes for power substations
    else:
        colors.append("black")
        sizes.append(10)

plt.figure(figsize=(12, 10))

nx.draw_networkx_edges(G, pos, edge_color="lightgray", alpha=0.7)

nx.draw_networkx_nodes(G, pos, node_color=colors, node_size=sizes)

plt.scatter([], [], c='gray', s=10, label='Road')
plt.scatter([], [], c='red', s=40, label='Hospital')
plt.scatter([], [], c='blue', s=60, label='Power Substation')
plt.legend(loc="upper right", title="Infrastructure Types")

plt.title("CascadeWatch Infrastructure Dependency Graph", fontsize=16)
plt.axis("equal") 
plt.axis("off")   
plt.tight_layout()
plt.show()
```
